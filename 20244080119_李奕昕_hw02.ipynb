### 任务 1：判断回文
忽略大小写和空格后，检查字符串正读和反读是否相同。

```python
# 第二个代码单元：按作业要求加入 Python 注释。

def is_palindrome(text):
    normalized = "".join(text.lower().split())
    return normalized == normalized[::-1]

test_cases = [
    "level",
    "T A C O cat",
    "Was it a car or a cat I saw",
    "hello",
]

for text in test_cases:
    print(f"{text!r} 是否为回文: {is_palindrome(text)}")
```

### 任务 2：判断字母顺序词
忽略空格，判断字符是否按字母顺序排列；允许字母重复。

```python
def is_abecedarian(text):
    letters = "".join(text.split())
    return all(
        letters[index] <= letters[index + 1]
        for index in range(len(letters) - 1)
    )

test_cases = ["adder", "beet", "dog", "cat", "abcd efgh xyz"]

for text in test_cases:
    print(f"{text!r} 是否按字母顺序排列: {is_abecedarian(text)}")
```

### 任务 3：移除字符串中的元音
移除英文字母 `a、e、i、o、u`，同时处理大写元音并保留其他字符。

```python
def remove_vowels(text):
    vowels = set("aeiouAEIOU")
    return "".join(character for character in text if character not in vowels)

test_cases = ["cat", "audio", "AEIOU", "rhythm"]

for text in test_cases:
    print(f"{text!r} 移除元音后: {remove_vowels(text)!r}")
```

### 任务 4：反转列表
返回列表的反转副本，不修改原列表；如果参数不是列表，则引发 `TypeError`。

```python
def list_reverse(values):
    if not isinstance(values, list):
        raise TypeError("list_reverse 的参数必须是列表")
    return values[::-1]

original = [1, 2, 3]
reversed_list = list_reverse(original)
print(f"原列表: {original}")
print(f"反转后的列表: {reversed_list}")

try:
    list_reverse((1, 2, 3))
except TypeError as error:
    print(f"非列表参数测试: {type(error).__name__}: {error}")
```

### 任务 5：判断序列是否按升序排列
只遍历序列一次，逐个比较相邻元素。空序列和单元素序列视为已排序。

```python
def is_sorted(seq):
    iterator = iter(seq)
    try:
        previous = next(iterator)
    except StopIteration:
        return True

    for current in iterator:
        if previous > current:
            return False
        previous = current

    return True

test_cases = [
    [1, 2, 2, 4],
    [1, 3, 2, 4],
    [],
    [5],
    ("a", "b", "c"),
]

for sequence in test_cases:
    print(f"{sequence!r} 是否按升序排列: {is_sorted(sequence)}")
```

### 任务 6：向量与标量相乘
向量使用数字元组表示。检查向量和标量类型，返回与输入向量长度相同的新元组。

```python
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("向量必须是元组")
    if any(type(value) not in (int, float) for value in t):
        raise TypeError("向量元素必须是整数或浮点数")
    if type(s) not in (int, float):
        raise TypeError("标量必须是整数或浮点数")
    return tuple(value * s for value in t)

print(f"(1, 2.5, -3) * 2 = {vec_scalar_mult((1, 2.5, -3), 2)}")
print(f"(4, 5) * 0.5 = {vec_scalar_mult((4, 5), 0.5)}")

for vector, scalar in [([1, 2], 2), ((1, "x"), 2), ((1, 2), "2")]:
    try:
        print(vec_scalar_mult(vector, scalar))
    except TypeError as error:
        print(f"错误测试: {type(error).__name__}: {error}")
```

### 任务 7：计算向量内积
输入必须是两个数字元组；长度不一致时引发 `ValueError`，结果以浮点数返回。

```python
import math

def vec_inner_product(vector_a, vector_b):
    if not isinstance(vector_a, tuple) or not isinstance(vector_b, tuple):
        raise TypeError("两个向量都必须是元组")
    if len(vector_a) != len(vector_b):
        raise ValueError("两个向量的长度必须相同")
    if any(type(value) not in (int, float) for value in vector_a + vector_b):
        raise TypeError("向量元素必须是整数或浮点数")
    return float(math.fsum(a * b for a, b in zip(vector_a, vector_b)))

print(f"(1, 2, 3) 与 (4, 5, 6) 的内积: {vec_inner_product((1, 2, 3), (4, 5, 6))}")
print(f"两个空向量的内积: {vec_inner_product((), ())}")

try:
    vec_inner_product((1, 2), (3,))
except ValueError as error:
    print(f"长度错误测试: {type(error).__name__}: {error}")

try:
    vec_inner_product([1, 2], (3, 4))
except TypeError as error:
    print(f"类型错误测试: {type(error).__name__}: {error}")
```

### 任务 8：验证矩阵
矩阵用元组的元组表示。每一行必须是元组，各行长度相同，所有元素必须是整数或浮点数。

```python
def check_valid_mx(matrix):
    if not isinstance(matrix, tuple):
        return False
    if any(not isinstance(row, tuple) for row in matrix):
        return False
    if matrix:
        row_length = len(matrix[0])
        if any(len(row) != row_length for row in matrix):
            return False
    return all(
        type(value) in (int, float)
        for row in matrix
        for value in row
    )

test_matrices = [
    ((1, 2), (3, 4)),
    ((1, 2), (3,)),
    ((1, "x"),),
    [[1, 2], [3, 4]],
    (),
]

for matrix in test_matrices:
    print(f"{matrix!r} 是否为有效矩阵: {check_valid_mx(matrix)}")
```

### 任务 9：矩阵与向量相乘
先验证矩阵和向量；矩阵列数必须等于向量长度。不合理的输入会引发相应异常。

```python
def mx_vec_mult(matrix, vector):
    if not check_valid_mx(matrix):
        raise ValueError("矩阵无效")
    if not isinstance(vector, tuple):
        raise TypeError("向量必须是元组")
    if any(type(value) not in (int, float) for value in vector):
        raise TypeError("向量元素必须是整数或浮点数")
    if matrix:
        column_count = len(matrix[0])
        if len(vector) != column_count:
            raise ValueError("向量长度必须等于矩阵的列数")
    elif vector:
        raise ValueError("空矩阵无法匹配非空向量")

    return tuple(
        sum(element * vector[index] for index, element in enumerate(row))
        for row in matrix
    )

matrix = ((1, 2, 3), (4, 5, 6))
vector = (7, 8, 9)
print(f"矩阵: {matrix}")
print(f"向量: {vector}")
print(f"矩阵与向量相乘的结果: {mx_vec_mult(matrix, vector)}")

for test_matrix, test_vector in [
    (((1, 2), (3, 4)), (1,)),
    (((1, 2), (3,)), (1, 2)),
    (((1, 2),), [1, 2]),
]:
    try:
        print(mx_vec_mult(test_matrix, test_vector))
    except (TypeError, ValueError) as error:
        print(f"错误测试: {type(error).__name__}: {error}")
```

### 任务 10：定义向量类 `Vector`
维度和坐标属于每个向量对象，因此使用实例属性。坐标保存为元组；不提供坐标时默认创建零向量。比较按坐标元组的字典序进行，不同维度不能比较或计算内积。

```python
import math

_MISSING = object()

class Vector:
    def __init__(self, dimension, elements=_MISSING):
        if type(dimension) is not int:
            raise TypeError("向量维度必须是整数")
        if dimension < 0:
            raise ValueError("向量维度不能为负数")

        if elements is _MISSING:
            elements = (0,) * dimension
        elif not isinstance(elements, (list, tuple)):
            raise TypeError("向量元素必须使用列表或元组提供")

        if any(type(value) not in (int, float) for value in elements):
            raise TypeError("向量元素必须是整数或浮点数")
        if len(elements) != dimension:
            raise ValueError("向量维度必须与元素个数相同")

        self.dimension = dimension
        self.elements = tuple(elements)

    def __repr__(self):
        return f"Vector({self.dimension}, {self.elements})"

    def _comparison_elements(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dimension != other.dimension:
            raise ValueError("不能比较维度不同的向量")
        return other.elements

    def __eq__(self, other):
        other_elements = self._comparison_elements(other)
        if other_elements is NotImplemented:
            return NotImplemented
        return self.elements == other_elements

    def __lt__(self, other):
        other_elements = self._comparison_elements(other)
        if other_elements is NotImplemented:
            return NotImplemented
        return self.elements < other_elements

    def __le__(self, other):
        other_elements = self._comparison_elements(other)
        if other_elements is NotImplemented:
            return NotImplemented
        return self.elements <= other_elements

    def __gt__(self, other):
        other_elements = self._comparison_elements(other)
        if other_elements is NotImplemented:
            return NotImplemented
        return self.elements > other_elements

    def __ge__(self, other):
        other_elements = self._comparison_elements(other)
        if other_elements is NotImplemented:
            return NotImplemented
        return self.elements >= other_elements

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("内积参数必须是 Vector 对象")
        if self.dimension != other.dimension:
            raise ValueError("计算内积的向量必须具有相同维度")
        return float(math.fsum(
            a * b for a, b in zip(self.elements, other.elements)
        ))

    def __mul__(self, other):
        if isinstance(other, Vector):
            if self.dimension != other.dimension:
                raise ValueError("相乘的向量必须具有相同维度")
            return Vector(
                self.dimension,
                tuple(a * b for a, b in zip(self.elements, other.elements)),
            )

        if type(other) in (int, float):
            return Vector(
                self.dimension,
                tuple(value * other for value in self.elements),
            )

        return NotImplemented

    def __rmul__(self, other):
        if type(other) in (int, float):
            return self * other
        return NotImplemented

vector_a = Vector(2, (2, 4))
vector_b = Vector(2, (2, 5))
vector_c = Vector(2, [4, 5])
zero_vector = Vector(3)

print(f"向量 A: {vector_a}")
print(f"默认初始化的零向量: {zero_vector}")
print(f"A == A: {vector_a == vector_a}")
print(f"A < B: {vector_a < vector_b}")
print(f"A <= B: {vector_a <= vector_b}")
print(f"B > A: {vector_b > vector_a}")
print(f"B >= A: {vector_b >= vector_a}")
print(f"A 与 C 的内积: {vector_a.dot(vector_c)}")
print(f"A * 2: {vector_a * 2}")
print(f"2 * A: {2 * vector_a}")
print(f"A * C（逐坐标相乘）: {vector_a * vector_c}")

for operation in [
    lambda: Vector(-1),
    lambda: Vector(2, (1,)),
    lambda: vector_a < Vector(3, (1, 2, 3)),
    lambda: vector_a.dot((1, 2)),
    lambda: vector_a.dot(Vector(3, (1, 2, 3))),
    lambda: vector_a * Vector(3, (1, 2, 3)),
]:
    try:
        print(operation())
    except (TypeError, ValueError) as error:
        print(f"异常测试: {type(error).__name__}: {error}")
```
