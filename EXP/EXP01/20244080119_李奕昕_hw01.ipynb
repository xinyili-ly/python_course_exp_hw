## 任务1：热身简单函数
1. say_hi()：打印 Hello, world!
2. goat_pad(s)：字符串前后拼接goat
3. print_n(s, n)：字符串打印n次，每行一条

```python
def say_hi():
    print("Hello, world!")

def goat_pad(s):
    res = "goat" + s + "goat"
    print(res)

def print_n(s, n):
    for _ in range(n):
        print(s)

# 测试
print("=== say_hi 测试 ===")
say_hi()
print("\n=== goat_pad 测试 ===")
goat_pad("test")
print("\n=== print_n 测试 ===")
print_n("Hi", 3)
```

## 任务2：欧几里得算法 求最大公约数 gcd(m,n)

```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m

# 测试三组
test_cases = [(2018, 2019), (1200, 300), (5040, 60)]
for a, b in test_cases:
    print(f"gcd({a}, {b}) = {gcd(a, b)}")

# 负数测试
print(f"\n负数测试 gcd(-12, 8) = {gcd(-12, 8)}")
```

## 任务3：近似计算欧拉数 e

```python
import math

def euler_limit(n):
    return (1 + 1 / n) ** n

def euler_infinite_sum(n):
    from math import factorial
    total = 0
    for k in range(n):
        total += 1 / factorial(k)
    return total

def euler_approx(eps):
    from math import factorial
    true_e = math.exp(1)
    sum_e = 0
    k = 0
    while abs(sum_e - true_e) > eps:
        sum_e += 1 / factorial(k)
        k += 1
    return sum_e

def print_euler_sum_table(n):
    for k in range(1, n + 1):
        val = euler_infinite_sum(k)
        print(f"k={k:2d} | sum = {val:.8f}")

# 测试
print("=== euler_limit 测试 n=10000 ===")
print(f"e ≈ {euler_limit(10000):.8f}\n")

print("=== euler_infinite_sum 测试 ===")
print(f"euler_infinite_sum(0) = {euler_infinite_sum(0)}")
print(f"euler_infinite_sum(1) = {euler_infinite_sum(1)}")
print(f"euler_infinite_sum(4) = {euler_infinite_sum(4):.6f}\n")

print("=== euler_approx 误差0.001 ===")
print(euler_approx(0.001))

print("\n=== 级数求和表 ===")
print_euler_sum_table(8)
```

## 任务4：测试整数是否为幂次

```python
def is_power_of_2(n):
    if n <= 0:
        return False
    while n % 2 == 0:
        n = n // 2
    return n == 1

def is_power(b, n):
    if n <= 0 or b <= 1:
        return False
    while n % b == 0:
        n = n // b
    return n == 1

# 测试
print("=== is_power_of_2 ===")
test_list = [1, 2, 4, 8, 7, 10]
for num in test_list:
    print(f"{num} 是否是2的幂: {is_power_of_2(num)}")

print("\n=== is_power(b,n) ===")
print(f"is_power(3,81) = {is_power(3, 81)}")
print(f"is_power(5,100) = {is_power(5, 100)}")
print(f"is_power(2,32) = {is_power(2, 32)}")
```
