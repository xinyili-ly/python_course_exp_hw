实验2：函数、条件、循环与递归

```python
# 任务1.分数等级判断函数
def score_level(score):
    if not (0 <= score <=100):
        raise ValueError("分数必须在0‑100之间")
    if 90 <= score <=100:
        return "优秀"
    elif 80 <= score <90:
        return "良好"
    elif 60 <= score <80:
        return "及格"
    else:
        return "不及格"

print("------分数等级测试------")
test_scores = [95,82,70,45]
for s in test_scores:
    print(f"分数：{s:>3} → 等级：{score_level(s)}")
```

```python
# 任务2.for、while循环求1‑100累加和
sum_for = 0
for i in range(1,101):
    sum_for += i

sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i += 1

print(f"for循环计算1‑100总和：{sum_for}")
print(f"while循环计算1‑100总和：{sum_while}")
```

```python
# 任务3.迭代阶乘、递归阶乘
def fact_iter(n):
    if n < 0:
        raise ValueError("输入需要为非负整数")
    res = 1
    for k in range(2, n+1):
        res *= k
    return res

def fact_recur(n):
    if n < 0:
        raise ValueError("输入需要为非负整数")
    if n == 0:
        return 1
    return n * fact_recur(n-1)

print("------阶乘结果对比------")
for n in [0,1,5,10]:
    print(f"n = {n:2d} | 迭代阶乘 = {fact_iter(n):<6d} | 递归阶乘 = {fact_recur(n):<6d}")
```

```python
# 任务4.summary统计函数，元组返回结果并解包
def summary(values):
    if not values:
        raise ValueError("输入列表不能为空")
    minv = min(values)
    maxv = max(values)
    avgv = sum(values)/len(values)
    return minv, maxv, avgv

vals = [10,20,30,40]
mi, ma, av = summary(vals)
print(f"输入列表：{vals}")
print(f"最小值 = {mi}, 最大值 = {ma}, 平均值 = {av:.2f}")
```

```python
# 任务5.sum_all(*args)可变参数求和
def sum_all(*args):
    return sum(args)

print(f"无参数调用sum_all()     → {sum_all()}")
print(f"单个参数sum_all(10)     → {sum_all(10)}")
print(f"多参数sum_all(1,2,3,4)  → {sum_all(1,2,3,4)}")
```

```python
# 任务1.zip构建学生成绩字典，items遍历输出
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

stu_dict = dict(zip(names,scores))
print("------学生姓名与成绩------")
for k,v in stu_dict.items():
    print(f"姓名：{k:<10} 成绩：{v:>3}")
```

```python
# 任务2.enumerate带序号遍历，序号从1开始
print("------带序号输出学生信息------")
for idx,(name,sc) in enumerate(zip(names,scores), start=1):
    print(f"序号{idx:2d} | 姓名：{name:<10} | 成绩：{sc:>3}")
```

```python
# 任务3.列表推导式筛选及格姓名，生成成绩平方列表
pass_names = [name for name,sc in zip(names,scores) if sc>=60]
score_sq = [s*s for s in scores]

print(f"及格学生姓名：{pass_names}")
print(f"成绩平方列表：{score_sq}")
```

```python
# 任务4.成绩统计：人数，平均分，最高最低，及格率
count = len(scores)
avg = sum(scores)/count
max_sc = max(scores)
min_sc = min(scores)
pass_cnt = sum(1 for s in scores if s>=60)
pass_rate = pass_cnt / count

print(f"总人数：{count}")
print(f"平均分：{avg:.2f}")
print(f"最高分：{max_sc}")
print(f"最低分：{min_sc}")
print(f"及格率：{pass_rate:.2%}")
```

实验4：文件读写、异常、模块


```python
# 任务1.查看工作目录，创建exp01_data文件夹
import os
import pickle
import json

print(f"当前工作目录：{os.getcwd()}")
data_dir = os.path.join(".","exp01_data")
os.makedirs(data_dir, exist_ok=True)
print(f"数据文件夹路径：{data_dir}")
```

```python
# 任务2.写入students.txt，读取文件并跳过表头
txt_path = os.path.join(data_dir,"students.txt")
content_text = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
with open(txt_path,"w",encoding="utf-8") as f:
    f.write(content_text)

with open(txt_path,"r",encoding="utf-8") as f:
    lines = f.readlines()

print("------读取到的有效行（跳过表头）------")
for line_idx,line in enumerate(lines[1:], start=2):
    line = line.strip()
    if not line:
        continue
    print(f"行{line_idx:2d}: {line}")

```

```python
# 任务3.解析每行学生数据，捕获各类异常，收集有效记录与错误日志
valid_records = []
error_log = []
with open(txt_path,"r",encoding="utf-8") as f:
    lines = f.readlines()

for line_idx,line in enumerate(lines[1:], start=2):
    line = line.strip()
    if not line:
        continue
    try:
        parts = line.split(",")
        if len(parts)!=2:
            raise ValueError("字段数量不对")
        name_part = parts[0].strip()
        score_str = parts[1].strip()
        if len(name_part)==0:
            raise ValueError("姓名为空")
        sc = int(score_str)
        if not (0<=sc<=100):
            raise ValueError(f"成绩{sc}不在0‑100")
        valid_records.append({"name":name_part,"score":sc})
    except ValueError as err:
        error_log.append({"line":line_idx,"line_content":line,"err_msg":str(err)})


print("====有效记录====")
print(json.dumps(valid_records,indent=2,ensure_ascii=False))

print("\n====错误记录日志====")
print(json.dumps(error_log,indent=2,ensure_ascii=False))
```

```python
# 任务4.生成report.txt，w覆盖写入，a追加写入
valid_num = len(valid_records)
invalid_num = len(error_log)
if valid_num>0:
    avg_score = sum(r["score"] for r in valid_records)/valid_num
    pass_num = sum(1 for r in valid_records if r["score"]>=60)
    pass_r = pass_num / valid_num
else:
    avg_score=0
    pass_r=0

report_path = os.path.join(data_dir,"report.txt")
# w模式覆盖
with open(report_path,"w",encoding="utf-8") as f:
    f.write(f"有效记录数:{valid_num}\n")
    f.write(f"无效记录数:{invalid_num}\n")
    f.write(f"平均分:{avg_score:.2f}\n")
    f.write(f"及格率:{pass_r:.2%}\n")

# a模式追加
with open(report_path,"a",encoding="utf-8") as f:
    f.write("=====追加(a模式)测试文本=====\n")

print(f"report.txt已写入，路径：{report_path}")
```

```python
# 任务5.pickle保存和读取对象
pkl_path = os.path.join(data_dir,"students.pkl")
with open(pkl_path,"wb") as f:
    pickle.dump(valid_records,f)

with open(pkl_path,"rb") as f:
    loaded_data = pickle.load(f)

print("pickle读取的数据：")
print(json.dumps(loaded_data,indent=2,ensure_ascii=False))
print(f"\n恢复的数据与原数据相等：{loaded_data == valid_records}")
```

实验5：类、继承、对象复制


```python
# 任务1.定义Student学生类，is_passed方法，__str__
import copy

class Student:
    def __init__(self,name,score):
        self.name = name
        self.score = score
    def is_passed(self):
        return self.score >= 60
    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"
```

```python
# 任务2.用实验4的有效记录创建Student实例
print("------Student对象实例------")
for rec in valid_records:
    s_obj = Student(rec["name"], rec["score"])
    print(f"{str(s_obj)}  | 是否及格：{s_obj.is_passed()}")

```

```python
# 任务3.GraduateStudent继承Student
class GraduateStudent(Student):
    def __init__(self,name,score,research_topic):
        super().__init__(name,score)
        self.research_topic = research_topic
    def __str__(self):
        return f"GraduateStudent(name={self.name}, score={self.score}, 研究方向={self.research_topic})"

print("------研究生子类测试------")
g_stu = GraduateStudent("LiMing",88,"大模型")
print(f"{str(g_stu)} | 是否及格：{g_stu.is_passed()}")
```

```python
# 任务4.直接赋值、copy.copy、copy.deepcopy对比

# 直接赋值
orig1 = {"name": "Alice", "scores": [85, 90]}
assign_obj = orig1
assign_obj["scores"][0] = 999
print(f"【直接赋值】原始对象：{orig1}")
print(f"【直接赋值】副本对象：{assign_obj}")

# 浅拷贝
orig2 = {"name": "Alice", "scores": [85, 90]}
shallow_obj = copy.copy(orig2)
shallow_obj["scores"][0] = 777
print(f"【copy.copy】原始对象：{orig2}")
print(f"【copy.copy】副本对象：{shallow_obj}")

# 深拷贝
orig3 = {"name": "Alice", "scores": [85, 90]}
deep_obj = copy.deepcopy(orig3)
deep_obj["scores"][0] = 555
print(f"【copy.deepcopy】原始对象：{orig3}")
print(f"【copy.deepcopy】副本对象：{deep_obj}")
```
